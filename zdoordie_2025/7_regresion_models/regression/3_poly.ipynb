```python
import numpy as np 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
X=6*np.random.rand(100,1)-3
X
```

```python
y=0.5 * X**2 +1.5*X+2 +np.random.rand(100,1)
y
```

```python
  
plt.scatter(X, y, marker='.')
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=43)
```

```python
from sklearn.linear_model import LinearRegression
regression=LinearRegression()
```

```python
regression.fit(X_train,y_train)
```

```python
from sklearn.metrics import r2_score

score=r2_score(y_test,regression.predict(X_test))
score
```

```python
plt.plot(X_train,regression.predict(X_train),color='r')
plt.scatter(X_train,y_train)
```

```python
from sklearn.preprocessing import PolynomialFeatures

poly=PolynomialFeatures(degree=2,include_bias=True)
```

```python
X_train_ploy=poly.fit_transform(X_train)
X_test_ploy=poly.fit_transform(X_test)
```

```python
X_train_ploy
```

```python
X_test_ploy
```

```python
regression=LinearRegression()

regression.fit(X_train_ploy,y_train)

ypred_poly=regression.predict(X_test_ploy)
print(ypred_poly)



score=r2_score(y_test,ypred_poly)
print(score)


```

```python
plt.scatter(X_train,regression.predict(X_train_ploy),marker='.',color='r')
plt.scatter(X_train,y_train)
```

```python
from sklearn.pipeline import Pipeline
def poly_regression(degree):
    X_new = np.linspace(-3, 3, 200).reshape(200, 1)

    poly_features = PolynomialFeatures(degree=degree, include_bias=True)
    lin_reg = LinearRegression()
    
    poly_regression = Pipeline([
        ("poly_features", poly_features),
        ("lin_reg", lin_reg)
    ])
    
    poly_regression.fit(X_train, y_train)
    y_pred_new = poly_regression.predict(X_new)

    # Plotting prediction line
    plt.plot(X_new, y_pred_new, "r-", label="Degree " + str(degree), linewidth=3)
    plt.plot(X_train, y_train, "b.", linewidth=3)
    plt.plot(X_test, y_test, "g.", linewidth=3)
    plt.legend(loc="upper left")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.axis([-4, 4, 0, 10])
    plt.show()
```

```python
poly_regression(3)
```
