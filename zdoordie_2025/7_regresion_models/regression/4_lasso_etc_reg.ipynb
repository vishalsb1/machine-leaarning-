```python
X=df_cp.drop('FWI',axis=1) #independent
y=df['FWI'] #dependent 
```

```python
from sklearn.model_selection import train_test_split

Xtrain,Xtest,ytrain,ytest=train_test_split(X,y,test_size=0.25,random_state=43)
```

```python
Xtrain.shape,Xtest.shape
```

```python
# feature selection using corr

df_cp.corr()
```

```python
plt.figure(figsize=(12,10))
corr=Xtrain.corr()
sns.heatmap(corr,annot=True)
```

```python
def extract_high_colinearity(dataset,threshold):
    correlated_col=set()
    corr_matrix=dataset.corr()

    for i in range(len(corr_matrix.columns)):
        for j in range(i):
            if abs(corr_matrix.iloc[i,j]>threshold):
                correlated_col.add(corr_matrix.columns[i])
            
    return correlated_col
```

```python
colm=extract_high_colinearity(Xtrain,0.85)
colm
```

```python
# so as their are highly correlated so lets drop them 

Xtrain.drop(colm,axis=1,inplace=True)
Xtest.drop(colm,axis=1,inplace=True)

Xtrain.shape,Xtest.shape
```

```python
# feature scaling and standardization

from sklearn.preprocessing import StandardScaler

scalar=StandardScaler()
```

```python
X_train_scaled=scalar.fit_transform(Xtrain)
X_test_scaled=scalar.transform(Xtest)
```

```python
# before standrdizaion
sns.boxplot(Xtrain)
```

```python
# after scaling - they come in an similar range 
sns.boxplot(X_train_scaled)
```

```python
# linear regression

from sklearn.linear_model import LinearRegression

regression=LinearRegression()
```

```python
regression.fit(X_train_scaled,ytrain)
```

```python
ypred=regression.predict(X_test_scaled)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
mae=mean_absolute_error(ytest,ypred)
mse=mean_squared_error(ytest,ypred)
score=r2_score(ytest,ypred)

mae,mse,score
```

```python
plt.scatter(ytest, ypred, color='blue', label="Predicted vs Actual")
plt.plot(ytest, ytest, 'r-', label="Ideal line (y=x)")
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.legend()
plt.show()
```

```python
# lasso regression 

from sklearn.linear_model import Lasso

lasso=Lasso()
lasso.fit(X_train_scaled,ytrain)
ypred=lasso.predict(X_test_scaled)

from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error

mae=mean_absolute_error(ytest,ypred)
mse=mean_squared_error(ytest,ypred)
score=r2_score(ytest,ypred)

print(mae,mse,score)
plt.scatter(ytest, ypred, color='blue', label="Predicted vs Actual")
plt.plot(ytest, ytest, 'r-', label="Ideal line (y=x)")
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.legend()
plt.show()
```

```python
# ridge regresiion

from sklearn.linear_model import Ridge

ridge=Ridge()
ridge.fit(X_train_scaled,ytrain)
ypred=ridge.predict(X_test_scaled)

from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error

mae=mean_absolute_error(ytest,ypred)
mse=mean_squared_error(ytest,ypred)
score=r2_score(ytest,ypred)

print(mae,mse,score)
plt.scatter(ytest, ypred, color='blue', label="Predicted vs Actual")
plt.plot(ytest, ytest, 'r-', label="Ideal line (y=x)")
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.legend()
plt.show()
```

```python
# elastic net 

from sklearn.linear_model import ElasticNet

en=ElasticNet()
en.fit(X_train_scaled,ytrain)
ypred=en.predict(X_test_scaled)

from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error

mae=mean_absolute_error(ytest,ypred)
mse=mean_squared_error(ytest,ypred)
score=r2_score(ytest,ypred)

print(mae,mse,score)
plt.scatter(ytest, ypred, color='blue', label="Predicted vs Actual")
plt.plot(ytest, ytest, 'r-', label="Ideal line (y=x)")
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.legend()
plt.show()
```
