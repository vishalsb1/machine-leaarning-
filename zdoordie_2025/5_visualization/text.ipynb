# matplotlib

```python
# basics
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.linspace(0,5,11)
y=x**2
```

```python
plt.plot(x,y)
plt.title("x vs y")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
```

```python
# subplots

plt.subplot(2,2,1)
plt.plot(y,x)

plt.subplot(2,2,4)
plt.plot(x,y)
```

```python
# object oriented way

# plot cya att plot

f=plt.figure()
axis1=f.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2=f.add_axes([0.2,0.4,0.3,0.3])
axis2.plot(y,x)
```

```python
# types of plot

plt.scatter(x,y)
```

```python
plt.hist(y)
```

```python
plt.boxplot(x)
```

```python
# saving plots

fig=plt.figure()
ax=fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("base_plot.png")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=sns.load_dataset('tips')
```

```python
df
```

```python
sns.displot(df["total_bill"],kde=True) #kde = kernal desnsity estimator that heps us to find the skweness 
```

```python
plt.subplot(1,2,1)
sns.histplot(df["total_bill"],kde=True) 
plt.subplot(1,2,2)
sns.histplot(df["tip"],kde=True) 
```

```python
sns.jointplot(x="total_bill",y="tip",data=df,kind="scatter")
```

```python
sns.jointplot(x="total_bill",y="tip",data=df,kind="kde")
```

```python
sns.jointplot(x="total_bill",y="tip",data=df,kind="hex")
```

```python
sns.jointplot(x="total_bill",y="tip",data=df,kind="reg")
```

```python
sns.pairplot(df,hue="sex")
```

```python
sns.rugplot(df)
```

# matrixplot


```python
# ------------------------ heatmap ------------------------

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
fl=sns.load_dataset('flights')
fl
```

```python
tips=sns.load_dataset('tips')
tips
```

```python
# to build heatmap coralations are beain calulated for preediction value except
tipscorr=tips[['total_bill','tip','size']]
```

```python
tipscorr
```

```python
tipscorr.corr()
```

```python
sns.heatmap(tipscorr.corr(),annot=True)
```

```python
# -----------------------clustermap----------------------

sns.clustermap(tipscorr.corr(),annot=True)
```

```python
# -----------------------pivot table heatmap----------------------

pvtfl=fl.pivot_table(values='passengers',index="month",columns="year")
pvtfl
```

```python
sns.heatmap(pvtfl)
```
